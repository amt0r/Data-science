# Графовий аналіз енергетичних мереж

## Завдання
Створити графову модель електромережі та дослідити її стійкість до відмов, використовуючи бібліотеку `NetworkX`. Порівняти графовий підхід із раніше використаними методами статистичного аналізу надійності. Візуалізувати графи мережі до та після симульованої відмови. Код має бути написаний у стилі об'єктно-орієнтованого програмування (ООП).

## Суть завдання
Мета завдання полягає у моделюванні електромережі як сукупності взаємопов'язаних вузлів (підстанцій, генераторів, споживачів) та ліній електропередач (ребер графа).
На відміну від чисто статистичного аналізу (де ми дивимося лише на загальну ймовірність виходу з ладу, наприклад, через експоненціальний розподіл), графовий аналіз дозволяє побачити **топологічну вразливість**. Якщо з ладу виходить ключова підстанція (вузол з великою кількістю зв'язків), мережа може розпастися на кілька ізольованих частин, що призведе до каскадного відключення (блекауту). Ми симулюємо таку відмову і візуалізуємо наслідки.

## Опис результатів
Після виконання коду ви отримаєте:
1. **Графік "До відмови"**: Відображає цілісну структуру електромережі, де всі вузли підключені один до одного.
2. **Графік "Після відмови"**: Показує стан мережі після видалення найбільш навантаженого (центрального) вузла. Ви побачите, як змінюється зв'язність — деякі вузли можуть стати повністю ізольованими, формуючи "острови" без живлення.
3. **Аналітика в консолі**: Виведення базових метрик графа (кількість вузлів, ребер) до та після аварії. Також виводиться порівняння: статистичний метод каже нам лише "з якою ймовірністю станеться аварія", тоді як графовий метод точно показує "які саме райони залишаться без світла після цієї аварії".

## Приклади реального використання
1. **Запобігання каскадним відключенням (Блекаутам)**: Інженери енергетичних компаній (наприклад, Укренерго) постійно будують графи своїх мереж. Вони штучно "вимикають" у симуляції кожну підстанцію по черзі, щоб знайти найслабші місця. Якщо вимкнення однієї лінії перевантажує сусідні (які теж вимикаються), починається каскадна аварія. Графовий аналіз дозволяє знайти, де саме потрібно побудувати резервні лінії передач.
2. **Оптимізація Smart Grid**: При інтеграції зеленої енергетики (сонячні панелі на дахах, вітряки) мережа стає децентралізованою. NetworkX використовується для розрахунку того, як краще маршрутизувати енергію від локальних панелей до сусідніх будинків при пошкодженні центральної магістралі.
